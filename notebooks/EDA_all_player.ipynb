```python
# --- 設定（ここを変えると定義が変わる） ---
STATCAST_PATH = data_dir() / "external" / "statcast" / "2025_regular.parquet"
EXCLUDE_PITCH_TYPES = ("UN", "PO")  # 持ち球に含めないコード
MIN_PITCHES_PITCHER = 1  # 投手全体の最低投球数（例: 100 にすると先発寄り）
MIN_PITCHES_PER_TYPE = 10  # その球種を「持ち球」とみなす最低投球数（例: 10）

# --- 読み込み ---
df = pl.read_parquet(STATCAST_PATH, columns=["pitcher", "pitch_type"])

df = df.filter(
    pl.col("pitch_type").is_not_null()
    & ~pl.col("pitch_type").is_in(EXCLUDE_PITCH_TYPES)
)

# 投手ごとの総投球数で母集団を絞る
pitcher_totals = df.group_by("pitcher").agg(pl.len().alias("n_pitches"))
eligible_pitchers = pitcher_totals.filter(
    pl.col("n_pitches") >= MIN_PITCHES_PITCHER
).select("pitcher")

df = df.join(eligible_pitchers, on="pitcher", how="inner")

# 球種ごとの投球数 → 最低投球数を満たす球種だけ残す（MIN_PITCHES_PER_TYPE > 1 のとき意味がある）
type_counts = (
    df.group_by("pitcher", "pitch_type")
    .agg(pl.len().alias("n"))
    .filter(pl.col("n") >= MIN_PITCHES_PER_TYPE)
)

# 投手ごとの持ち球数
repertoire = type_counts.group_by("pitcher").agg(
    pl.col("pitch_type").n_unique().alias("n_pitch_types")
)

# リーグ全体の要約
summary = repertoire.select(
    pl.len().alias("n_pitchers"),
    pl.col("n_pitch_types").mean().alias("mean"),
    pl.col("n_pitch_types").median().alias("median"),
    pl.col("n_pitch_types").quantile(0.25).alias("q25"),
    pl.col("n_pitch_types").quantile(0.75).alias("q75"),
    pl.col("n_pitch_types").min().alias("min"),
    pl.col("n_pitch_types").max().alias("max"),
)
print(summary)

# 分布（何球種の投手が何人いるか）
dist = repertoire["n_pitch_types"].value_counts().sort("n_pitch_types")
print(dist)

# 多球種・少球種の確認用（上位・下位）
print(repertoire.sort("n_pitch_types", descending=True).head(10))
print(repertoire.sort("n_pitch_types").head(10))
```

```python
detail = (
    type_counts.sort("pitcher", "n", descending=[False, True])
    .group_by("pitcher")
    .agg(
        pl.col("pitch_type").alias("pitch_types"),
        pl.col("n").alias("pitches_per_type"),
    )
)
# detail.filter(pl.col("pitch_types").list.len() >= 6).head(5)
```
